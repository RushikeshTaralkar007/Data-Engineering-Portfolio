```python
import pandas as pd
from kafka import KafkaProducer
from time import sleep
from json import dumps
import json
```

```python
producer = KafkaProducer(bootstrap_servers=['16.4.38.168:9092'],
                         value_serializer=lambda x:
                         dumps(x).encode('utf-8'))
```

```python
producer.send('demo_rushi', value={'Name':'Rushi Taralkar'})
```

```python
df = pd.read_csv("nifty50.csv")
```

```python
df.head()
```

```python
while True:
    dict_stock = df.sample().to_dict(orient = "records")[0]
    producer.send("demo_rushi", value = dict_stock)
```
